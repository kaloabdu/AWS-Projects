# Student Score Data Processing
This project loads student scores, cleans invalid records using a function, and analyzes valid scores with NumPy.

```python
import pandas as pd
import numpy as np
from io import StringIO

# Create the raw student dataset
csv_data = """student,score
Abel,85
Sara,92
Mimi,78
Dawit,absent
Hana,105
Yonas,-5
Liya,67
Kebede,
Nati,45
"""

# Load the data
data = pd.read_csv(StringIO(csv_data))

print("Original dataset:")
display(data)
```

```python
def clean_scores(data):
    """Keep only numeric scores between 0 and 100."""
    data = data.copy()

    # Convert invalid text to missing values
    data["score"] = pd.to_numeric(data["score"], errors="coerce")

    # Remove missing scores
    data = data.dropna(subset=["score"])

    # Keep scores from 0 to 100
    data = data[data["score"].between(0, 100)]

    return data


# Clean the original dataset
cleaned_data = clean_scores(data)

print("Cleaned dataset:")
display(cleaned_data)
```

```python
# Convert the cleaned scores into a NumPy array
scores = cleaned_data["score"].to_numpy()

# Calculate statistics using NumPy
average_score = np.mean(scores)
highest_score = np.max(scores)
lowest_score = np.min(scores)
passing_students = np.sum(scores >= 50)

# Display the results
print("NumPy Analysis")
print("--------------------")
print("Average score:", round(average_score, 2))
print("Highest score:", highest_score)
print("Lowest score:", lowest_score)
print("Number of passing students:", passing_students)
```
